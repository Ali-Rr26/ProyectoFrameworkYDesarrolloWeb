## Documento de Requerimientos del Sistema (DRS) - Fase 2: Arquitectura y API Robusta

Aquí tienes el bloque de instrucciones para la **Fase 2**, enfocado en cumplir con el Tema 3 de la planificación: la integración segura de persistencia de datos relacionales con PostgreSQL.

Este *prompt* guía a la IA para que elimine los datos en memoria y conecte el sistema a una base de datos real, manteniendo intacta la comunicación con el frontend en React.

### Instrucciones de Ejecución para la IA (Fase 2)

Copia y pega este texto en tu IDE (como Windsurf) para generar esta fase:

```markdown
> **Objetivo:** Implementar la persistencia de datos relacionales utilizando PostgreSQL, reemplazando completamente los datos semilla (mock) en memoria sin alterar las firmas de las funciones de los controladores.
> 
> **Paso 1: Infraestructura (Docker)**
> Crea un archivo `docker-compose.yml` en la raíz de `Unidad2_Backend/multicatalogo-backend` que levante una instancia de PostgreSQL 16. Configura variables de entorno básicas (usuario, contraseña, base de datos `multicatalogo`).
> 
> **Paso 2: Conexión a la Base de Datos (`config/db.go`)**
> Crea un paquete `config` con un archivo `db.go`. Utiliza el driver `pgx` (o `database/sql` estándar) para establecer un pool de conexiones a PostgreSQL. Asegúrate de implementar el cierre de la conexión al finalizar la aplicación y manejo de variables de entorno mediante un archivo `.env`.
> 
> **Paso 3: Estructura Relacional (Script SQL)**
> Crea un archivo `database/init.sql` con la definición de las tablas:
> - `usuarios`: id, email, password (texto plano temporalmente), rol.
> - `productos`: id, nombre, descripcion, precio, categoria, img, galeria (puede ser un tipo ARRAY o JSONB).
> - `referidos`: id, nombre, nivel, ventas, parent_id (clave foránea a la misma tabla para construir el árbol).
> Incluye sentencias `INSERT` para poblar la base de datos exactamente con los mismos datos que existen actualmente en los arreglos `usuarios`, `productos` y `red` del archivo `repository.go`.
> 
> **Paso 4: Refactorización de la Capa de Datos (`repository/repository.go`)**
> Modifica el archivo `repository.go`:
> 1. Elimina las variables globales `usuarios`, `productos` y `red`.
> 2. Inyecta o utiliza la conexión a la base de datos creada en el Paso 2.
> 3. Modifica `BuscarUsuarioPorCredenciales` para que ejecute `SELECT * FROM usuarios WHERE email = $1 AND password = $2`.
> 4. Modifica `ObtenerProductos` y `ObtenerProductoPorID` para ejecutar sus respectivos `SELECT`.
> 5. Modifica `ObtenerRed`: Construye la jerarquía multinivel consultando la tabla `referidos` (puedes usar una consulta recursiva CTE o reconstruir el árbol en memoria a partir de una lista plana).

```

### ¿Cómo visualizar el avance en pantalla?

Una vez que la IA termine y levantes tu contenedor de Docker:

1. Abre tu base de datos (con DBeaver, pgAdmin o terminal) y realiza un `UPDATE productos SET precio = 99.99 WHERE id = 1;`.
2. En la aplicación React, navega a la vista de `/tienda` o `/catalogo` y recarga la página.


3. El "Serum Revitalizante" ahora reflejará el nuevo precio extraído en tiempo real desde PostgreSQL. La persistencia es un hecho.

Valida que los datos viajan correctamente desde la base de datos al frontend.
